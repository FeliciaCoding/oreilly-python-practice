### Floating-Point Numbers Are Approximations

```python
amount = 112.1234567899
```

### Class Decimal from the Python Standard Library's decimal Module

### Importing the Decimal Class from the decimal Module

```python
from decimal import Decimal
```

```python
Decimal('0.1') + Decimal('0.2') == Decimal('0.3')
```

### Creating Decimals

```python
Decimal('19.99')     # correct
Decimal(19.99)       # WRONG — inherits float's imprecision
Decimal(19)          # fine — ints convert exactly, no fractional part to lose
```

```python
x = Decimal('10.5')
```

```python
x
```

```python
print(f'{x:.20f}')
```

```python
y = Decimal(2)  # OK because integers are represented exactly
```

```python
y
```

```python
print(f'{y:.20f}')
```

```python
print(y)
```

### Compound-Interest Problem Requirements Statement

### Calculating Compound Interest

```python
while 1: 
    principal = Decimal (input("Entering the original amount : "))
    rate = Decimal (input("Entering the interest rate : "))
    if (principal <= 0): 
        print("principal cannot below 1")
        continue 

    years = int(input("Entering how many years : "))
    if (years < 0): 
        print("year cannot be negative")
        continue


for year in range(years +1): 
    result = principle *  (1 + rate) ** year 

```

```python
principal = Decimal('1000.00')
```

```python
rate = Decimal('0.05')
```

```python
for year in range(1, 11):
    amount = principal * (1 + rate) ** year
    print(f'{year:>2}{amount:>10.2f}   {amount}')
```
