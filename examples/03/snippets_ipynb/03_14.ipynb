# 3.14 Built-In Function range: A Deeper Look

```python
# range (start, end(not included), move)
for number in range(5, 10):
    print(number, end=' ')
```

```python
for number in range(0, 10, 2):
    print(number, end=' ')
```

```python
for number in range(10, 0, -2):
    print(number, end=' ')
```

#### Python doesn't error on this — it just recognizes the range is empty and produces zero iterations.

```python
for number in range(10, 0, 2):
    print(number, end=' ')
```
