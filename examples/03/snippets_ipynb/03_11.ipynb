### Execution Phases

### Initialization Phase

```python
total = 0
avg = 0
count  = 0
grades = [98, 45, 84, 34, 54, 64, 78, 89]
```

### Processing Phase

```python
for grade in grades:
    total += grade 
    count += 1
avg = total / count
```

### Termination Phase

```python
print("Average grade is : ", avg)
```
