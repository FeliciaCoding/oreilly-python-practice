# .strip(chars)
#### removes any matching characters from both the left and right ends of a string, stopping on each side when it reaches a character not in chars. It does not remove matching characters from the middle.

```python
'www.example.com'.strip('cmowza.')
```

```python
'www.example.com'.strip('w.com')
```

# .removeprefix(str) v.s. .removesuffix(str)

```python
s = 'www.example.com'
s = s.removeprefix('www.')
s = s.removesuffix('.com')

print(s)  # example
```
