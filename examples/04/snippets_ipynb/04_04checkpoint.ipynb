### Checkpoint 2 Snippets

**(IPython Session)** Use a `for` statement, `randint` and a conditional expression to simulate 20 coin flips, displaying H for heads and T for tails, each separated by a space.

```python
import random

for flip in range(20): 
    result = random.randint(0,1)
    if (result == 0):
        print("H", end=' ')
    else :
        print("T", end=" ")
```
