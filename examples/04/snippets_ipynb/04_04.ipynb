# 4.4 Random-Number Generation

### Rolling a Six-Sided Die

```python
import random
```

```python
for roll in range(10):
    print(random.randint(1, 6), end=' ')
```

```python
for roll in range(10):
    print(random.randint(1, 6), end=' ')
```

```python
import secrets
secrets.token_hex(16)
```

# secrets
for cryptographic purposes (passwords, tokens, keys), never use a standard seeded PRNG — use a cryptographically secure generator (e.g., secrets in Python, not random), since predictable seeds make those systems exploitable.

```python
secrets.randbelow(100)
```

```python
secrets.randbelow(100)
```

### Seeding the Random-Number Generator for Reproducibility

```python
random.seed(32)
```

```python
for roll in range(10):
    print(random.randint(1, 6), end=' ')
```

```python
for roll in range(10):
    print(random.randint(1, 6), end=' ')
```

### same "random" results every run



```python
random.seed(32)
```

```python
for roll in range(10):
    print(random.randint(1, 6), end=' ')
```

```python
random.seed(32)

for roll in range(10):
    print(random.randint(1,6), end=' ')
```
