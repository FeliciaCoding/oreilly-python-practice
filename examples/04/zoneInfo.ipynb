# ZoneInfo v.s. datetime
datetime → date + clock time
ZoneInfo → which time zone that clock time belongs to

```python
from datetime import datetime
from zoneinfo import ZoneInfo

tz = ZoneInfo("Europe/Zurich")

summer = datetime(2026, 7, 1, 12, 0, tzinfo=tz)
winter = datetime(2026, 1, 1, 12, 0, tzinfo=tz)

print(summer)  # ...+02:00
print(winter)  # ...+01:00
```

```python
datetime.now(ZoneInfo("Europe/Zurich"))
```
