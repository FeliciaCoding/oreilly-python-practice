# match ... case ...

```python
grade = "C"

match grade:
    case 'A':
        print("a")
    case 'B':
        print("b")
    case _:
        print("default")
```

```python
gpa = 4

match gpa:
    case 4:
        print("A")
    case 3:
        print("B")
    case 2:
        print("C")
    case 1:
        print("D")     
    case _:
        print("invalide")           
```

# While loop

```python
"""
char grade = 'C';
while (grade < 'Z') {
    grade++;
}
std::cout << grade;
"""

while grade < 'Z' :
    grade += 1
    print (grade)
```

```python
grade = 'C'
while grade < 'Z':
    grade = chr(ord(grade) + 1)
print(grade)
```
